# Bar chart

```python
# Use : used to shows the discrete values of categorical data

food_items = ["Idily","Dosa","chapati","Puri"]
like_count = [5,7,13,18]

plt.figure(figsize=(7,5))
#plt.bar(food_items,like_count)
plt.bar(food_items,like_count,color = ["brown","grey","yellow","orange"])

for i, count in enumerate(like_count):
    plt.text(i, count+0.2, count, ha = "center") # plt.text(x_index,y_index,value, pos)
    
plt.xlabel("food items")
plt.ylabel("Number of likes")
```

# Histogram 

```python
# Use: Histogram works on continous data
# Histogram is popularly used to find the shape of the data

import numpy as np
data = np.random.randn(1000) # normal distributed data : -3 to 3
plt.figure(figsize = (3,2))
plt.hist(data, edgecolor = 'black',bins = 30)
plt.title("Histogram Chart")
plt.xlabel("X-data")
plt.ylabel("Y-label")
plt.show()
```

# Pie Chart

```python
# Use : used to represent less numbers categorical data
# To represent the data in form of percentage/proposition
# in pie chart we shows percentage of various categories
mobile_company = ["Realme","Samsung","Nokia","Lava","Apple","IQoo","OnePlus"]
market_share = [12,18,8,10,10,22,20]

plt.figure(figsize = (7,4))
plt.pie(market_share,labels = mobile_company, autopct = "%1.1f%%")
plt.title("Mobile market share")
plt.savefig("ds20_piechart.jpg")
plt.show()
```

# donut chart

```python
# use : same as pie chart
# donut chart is like pie chart only but have space at the center
# this space can be used to add extra information

mobile_company = ["Realme","Samsung","Nokia","Lava","Apple","IQoo","OnePlus"]
market_share = [12,18,8,10,10,22,20]

plt.figure(figsize = (7,4))
plt.pie(market_share,labels = mobile_company, autopct = "%1.1f%%", wedgeprops = dict(width = 0.6))
plt.text(0,0,"Sales",ha = "center")
plt.title("Mobile market share")
plt.show()
```

# Method-1: using plt.subplot(): good for less number of plots

```python
# plt.subplot(nrows, ncols,index)
mobile_company = ["Realme","Samsung","Nokia","Lava","Apple","IQoo","OnePlus"]
market_share = [12,18,8,10,10,22,20]

plt.figure(figsize = (8,6))

plt.subplot(2,3,1)
plt.pie(market_share, labels = mobile_company)

plt.subplot(2,3,2)
plt.pie(market_share, labels = mobile_company)

plt.subplot(2,3,3)
plt.pie(market_share, labels = mobile_company)

plt.subplot(2,3,4)
plt.pie(market_share, labels = mobile_company)

plt.subplot(2,3,5)
plt.pie(market_share, labels = mobile_company)

plt.subplot(2,3,6)
plt.pie(market_share, labels = mobile_company)
plt.tight_layout() # use this if charts are overlapping
plt.show()
```

```python
import matplotlib.pyplot as plt

mobile_company = ["Realme", "Samsung", "Nokia", "Lava", "Apple", "IQoo", "OnePlus"]
market_share = [12, 18, 8, 10, 10, 22, 20]

plt.figure(figsize=(12, 8))

# 1. Pie chart
plt.subplot(2, 3, 1)
plt.pie(market_share, labels=mobile_company, autopct='%1.1f%%')
plt.title("Market Share")

# 2. Horizontal bar chart
plt.subplot(2, 3, 2)
plt.barh(mobile_company, market_share)
plt.title("Horizontal Bar Chart")

# 3. Box plot
plt.subplot(2, 3, 3)
plt.boxplot(market_share)
plt.title("Box Plot")

# 4. Histogram
plt.subplot(2, 3, 4)
plt.hist(market_share, bins=5, edgecolor='black')
plt.title("Histogram")

# 5. Scatter plot
plt.subplot(2, 3, 5)
plt.scatter(range(len(mobile_company)), market_share)
plt.xticks(range(len(mobile_company)), mobile_company, rotation=45)
plt.title("Scatter Plot")

# 6. Vertical bar chart
plt.subplot(2, 3, 6)
plt.bar(mobile_company, market_share)
plt.xticks(rotation=45)
plt.title("Vertical Bar Chart")

plt.tight_layout()
plt.show()
```

# 3D charts in matplotlib

```python
import numpy as np

z = np.linspace(0,10,100)
x = np.sin(z)
y = np.cos(z)

fig = plt.figure()
ax = fig.add_subplot(111, projection = '3d')
ax.plot(x,y,z)
```
